# Big Data Processing with PySpark

## Objective

The objective of this section is to compare data processing performance between Pandas and PySpark using the NYC Taxi dataset.

The analysis focuses on:

- Data loading performance
- Data aggregation performance
- Scalability
- Big Data readiness

```python
import pandas as pd
import time 

from pyspark.sql import SparkSession
from pyspark.sql.functions import count
```

```python
spark = SparkSession.builder\
    .appName('Smart City Analytics')  \
    .getOrCreate()
```

```python
spark
```

```python
start = time.time()

df_pandas = pd.read_csv(
    '../data/raw/NYC.csv'
)

end = time.time()

pandas_load_time = end - start

print(
    f'Pandas load time : {pandas_load_time:.2f} second'
)
```

```python
start = time.time()

df_spark = spark.read.csv(
    '../data/raw/NYC.csv',
    header= True,
    inferSchema=True 
)


end = time.time()

spark_load_time = end - start

print(
    f'PySpark Lead Time: {spark_load_time:.2f} seconds'
)
```

```python
df_pandas.shape
```

```python
print(
    (
        df_spark.count(),
        len(df_spark.columns)
    )
)
```

```python
start = time.time()

pandas_groupby = (
    df_pandas
    .groupby("vendor_id")
    .size()
)

end = time.time()

pandas_group_time = end - start

print(pandas_groupby)
print(
    f"Pandas GroupBy Time: {pandas_group_time:.4f}"
)
```

```python
start = time.time()

spark_groupby = (
    df_spark
    .groupBy("vendor_id")
    .agg(
        count("*")
        .alias("trip_count")
    )
)

spark_groupby.show()

end = time.time()

spark_group_time = end - start

print(
    f"PySpark GroupBy Time: {spark_group_time:.4f}"
)
```

```python
comparison = pd.DataFrame({
    "Operation": [
        "Load Dataset",
        "GroupBy Vendor"
    ],
    "Pandas (sec)": [
        pandas_load_time,
        pandas_group_time
    ],
    "PySpark (sec)": [
        spark_load_time,
        spark_group_time
    ]
})

comparison
```

# Big Data Processing Conclusion

## Findings

The performance comparison between Pandas and PySpark produced mixed results.

### Data Loading

PySpark loaded the dataset faster than Pandas.

### Aggregation

Pandas significantly outperformed PySpark for the vendor aggregation task.

### Interpretation

The NYC Taxi dataset contains approximately 1.4 million records, which remains manageable for in-memory processing with Pandas.

For relatively small analytical workloads, Pandas can outperform PySpark due to lower computational overhead.

However, PySpark becomes advantageous when processing very large datasets that exceed the memory capacity of a single machine.

## Business Value

PySpark provides scalability and distributed computing capabilities, making it suitable for enterprise-scale Big Data environments.
